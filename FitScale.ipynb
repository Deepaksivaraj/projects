```python
weight = float(input("Enter your weight:"))
unit = input("Kilogram or Pounds (K or L)")

if unit == "K" or unit =="k":
    weight = weight * 2.205
    unit = "lbs"
    print(f"The weight of lbs is {weight} {unit}")

elif unit == "L" or unit == "l":
    weight = weight / 2.205
    unit = "kgs"
    print(f"The weight of kgs is {weight} {unit}")

else:
    print(f"{unit} is not valid")
```
